Como el pipeline declarativo se despliega como recurso aparte, aqui materializamos su **misma logica de negocio** con las herramientas batch que ya dominamos (Modulos 2 y 3) — para tener resultados reales dentro del notebook, sin esperar a desplegar el Pipeline en el workspace. Estas son las tablas **Gold** de referencia para el resto del curso.


<div style="box-sizing:border-box;width:100%;background:#0c1a2e;border:1px solid #1e3a5f;border-left:4px solid #2563eb;border-radius:6px;margin:10px 0;overflow:hidden;">

    <tr>
      <td style="width:78px;padding:8px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">QUE HACE</td>
      <td style="padding:8px 16px 6px 0;color:#e2e8f0;font-size:13px;">Calcula y persiste dos tablas Gold: <code style="color:#38bdf8;">resumen_embudo_semanal</code> (agregado semanal del embudo) y <code style="color:#38bdf8;">clientes_riesgo_fraude</code> (features de riesgo por cliente).</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">PARA QUE</td>
      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">resumen_embudo_semanal</code> agrega el detalle diario de <code style="color:#38bdf8;">silver.resumen_embudo_diario</code> (Modulo 2) al grano que de verdad consume un dashboard ejecutivo. <code style="color:#38bdf8;">clientes_riesgo_fraude</code> resume por cliente las senales de riesgo de <code style="color:#38bdf8;">silver.transacciones</code> (Modulo 3) — la tabla que el Modulo 6 usara directamente como insumo de Feature Engineering para el modelo de fraude.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">F.date_trunc("week", "fecha")</code> agrupa por semana calendario; la segunda tabla combina <code style="color:#38bdf8;">groupBy</code> + agregaciones condicionales (<code style="color:#38bdf8;">F.sum(F.col(...).cast("int"))</code>) para contar transacciones fraudulentas y con pais distinto, siguiendo el mismo patron de agregacion del Modulo 2.</td>
    </tr>
  </table>
</div>


```python
# GOLD 1: resumen del embudo a nivel semanal, por segmento y canal
df_gold_embudo = (
    spark.table(tbl(SCHEMA_SILVER, "resumen_embudo_diario"))
    .withColumn("semana", F.date_trunc("week", "fecha"))
    .groupBy("semana", "segmento", "canal_origen", "tipo_evento")
    .agg(F.sum("sesiones").alias("sesiones"))
)

(
    df_gold_embudo.write.format("delta")
    .mode("overwrite").option("overwriteSchema", "true")
    .saveAsTable(tbl(SCHEMA_GOLD, "resumen_embudo_semanal"))
)
print(f"[OK] {tbl(SCHEMA_GOLD, 'resumen_embudo_semanal')} — {spark.table(tbl(SCHEMA_GOLD,'resumen_embudo_semanal')).count():,} registros")

# GOLD 2: features de riesgo de fraude por cliente — insumo directo del Modulo 6
df_clientes = spark.table(tbl(SCHEMA_BRONZE, "clientes")).select("cliente_id", "segmento", "pais", "fecha_registro")

df_gold_riesgo = (
    spark.table(tbl(SCHEMA_SILVER, "transacciones"))
    .groupBy("cliente_id")
    .agg(
        F.count("*").alias("num_transacciones"),
        F.sum(F.col("es_fraude").cast("int")).alias("num_transacciones_fraude"),
        F.avg("monto_total").alias("monto_promedio"),
        F.max("monto_total").alias("monto_maximo"),
        F.sum((F.col("pais_ip_transaccion") != F.col("pais_cliente")).cast("int")).alias("num_pais_distinto"),
    )
    .withColumn("tasa_fraude_historica", F.round(F.col("num_transacciones_fraude") / F.col("num_transacciones"), 4))
    .join(broadcast(df_clientes), on="cliente_id")
)

(
    df_gold_riesgo.write.format("delta")
    .mode("overwrite").option("overwriteSchema", "true")
    .saveAsTable(tbl(SCHEMA_GOLD, "clientes_riesgo_fraude"))
)
print(f"[OK] {tbl(SCHEMA_GOLD, 'clientes_riesgo_fraude')} — {spark.table(tbl(SCHEMA_GOLD,'clientes_riesgo_fraude')).count():,} registros")

display(spark.table(tbl(SCHEMA_GOLD, "clientes_riesgo_fraude")).orderBy(F.desc("tasa_fraude_historica")).limit(10))
```
