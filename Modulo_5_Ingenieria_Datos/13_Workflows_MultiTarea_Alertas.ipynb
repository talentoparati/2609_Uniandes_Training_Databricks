El Job de la Sesion 11 era lineal (1 &rarr; 2). El pipeline real de este curso tiene **dos transformaciones independientes** que pueden correr al mismo tiempo — el embudo de carrito y las features de riesgo de fraude no dependen una de la otra, solo comparten el mismo punto de partida. Modelamos eso como un DAG con una rama que se **bifurca y se vuelve a unir**:

```
1. validar_entorno
       |
       +-----------------+
       v                 v
2. calcular_embudo   3. calcular_riesgo_fraude     <- corren EN PARALELO
       +-----------------+
       v
4. validar_calidad_gold
       v
5. notificar_resultado
```

Databricks Free Edition limita a **5 task runs concurrentes por cuenta** — con solo 2 tasks corriendo en paralelo en este DAG, el Job queda comodamente dentro de ese limite.


<div style="box-sizing:border-box;width:100%;background:#0c1a2e;border:1px solid #1e3a5f;border-left:4px solid #2563eb;border-radius:6px;margin:10px 0;overflow:hidden;">

      <td style="padding:8px 16px 6px 0;color:#e2e8f0;font-size:13px;">Crea un Job de 5 tasks: 1 inicial, 2 en paralelo (fan-out), 1 de validacion que espera a ambas (fan-in) y 1 de notificacion final — con schedule diario pausado.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">PARA QUE</td>
      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;">Las tasks 2 y 3 declaran la <strong style="color:#e2e8f0;">misma dependencia</strong> (<code style="color:#38bdf8;">validar_entorno</code>) pero no dependen entre si — el scheduler de Databricks las lanza en paralelo apenas la task 1 termina. La task 4 declara dependencia de <strong style="color:#e2e8f0;">ambas</strong>, asi que espera a que las dos terminen antes de arrancar. Es el patron clasico de fan-out/fan-in para orquestar transformaciones independientes.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;">Cada elemento de la lista <code style="color:#38bdf8;">tasks</code> es un <code style="color:#38bdf8;">jobs.Task</code> con su <code style="color:#38bdf8;">task_key</code> y, cuando aplica, una lista de <code style="color:#38bdf8;">jobs.TaskDependency</code>. <code style="color:#38bdf8;">jobs.CronSchedule</code> con <code style="color:#38bdf8;">pause_status=PAUSED</code> deja el horario configurado sin activarlo — se activa manualmente cuando el equipo decide pasar a produccion.</td>
    </tr>
  </table>

````python
job_multi_tarea = w.jobs.create(
    name=f"[{ENTORNO}] ecommerce_pipeline_completo",
    tasks=[
        jobs.Task(
            task_key="validar_entorno",
            notebook_task=jobs.NotebookTask(notebook_path=f"{BASE_PATH}/00_Setup/00_Configuracion"),
        ),
        jobs.Task(
            task_key="calcular_embudo",
            depends_on=[jobs.TaskDependency(task_key="validar_entorno")],
            notebook_task=jobs.NotebookTask(
                notebook_path=f"{BASE_PATH}/Modulo_2_Procesamiento_Spark/05_Particionamiento_SparkUI_ShuffleTuning.ipynb"
            ),
        ),
        jobs.Task(
            task_key="calcular_riesgo_fraude",
            depends_on=[jobs.TaskDependency(task_key="validar_entorno")],
            notebook_task=jobs.NotebookTask(
                notebook_path=f"{BASE_PATH}/Modulo_5_Ingenieria_Datos/12_Pipeline_Medallion_BronzeSilverGold.ipynb"
            ),
        ),
        jobs.Task(
            task_key="validar_calidad_gold",
            depends_on=[
                jobs.TaskDependency(task_key="calcular_embudo"),
                jobs.TaskDependency(task_key="calcular_riesgo_fraude"),
            ],
            notebook_task=jobs.NotebookTask(notebook_path=f"{BASE_PATH}/Modulo_4_Gobernanza_UnityCatalog/10_Linaje_Auditoria.ipynb"),
        ),
        jobs.Task(
            task_key="notificar_resultado",
            depends_on=[jobs.TaskDependency(task_key="validar_calidad_gold")],
            notebook_task=jobs.NotebookTask(notebook_path=f"{BASE_PATH}/00_Setup/00_Configuracion"),
        ),
    ],
    schedule=jobs.CronSchedule(
        quartz_cron_expression="0 0 5 * * ?",  # 5:00 AM todos los dias
        timezone_id="America/Bogota",
        pause_status=jobs.PauseStatus.PAUSED,  # PAUSED en el curso — se activa en produccion
    ),
    tags={"curso": "uniandes-databricks", "modulo": "5"},
)

print(f"[OK] Job multi-tarea creado — job_id = {job_multi_tarea.job_id}")
print(f"5 tasks: 1 inicial -> 2 en paralelo -> 1 de validacion -> 1 de notificacion")
print(f"Schedule: diario 5:00 AM America/Bogota (PAUSADO por defecto)")
print(f"URL: {w.config.host}/jobs/{job_multi_tarea.job_id}")
````

      <td style="padding:8px 16px 6px 0;color:#e2e8f0;font-size:13px;">Calcula la tasa de fraude diaria de los ultimos 7 dias y evalua si el dia mas reciente supera un umbral de alerta.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">PARA QUE</td>
      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;">Un Job que corre a las 5 AM no sirve de mucho si nadie se entera cuando algo sale mal. Este query es exactamente el que alimentaria una <strong style="color:#e2e8f0;">Databricks SQL Alert</strong>: avisar cuando la tasa de fraude diaria se dispare — la senal que el Modulo 6 va a aprender a predecir con anticipacion.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;">Agregacion SQL estandar (<code style="color:#38bdf8;">GROUP BY fecha</code>) seguida de una comparacion simple en Python contra <code style="color:#38bdf8;">UMBRAL_ALERTA_PCT</code>. En produccion, esta misma logica de umbral vive dentro de la condicion de la Alert, no en un <code style="color:#38bdf8;">if</code> de Python.</td>
    </tr>
  </table>
</div>


````python
# Query que alimentaria la alerta — tasa de fraude del dia mas reciente
df_alerta = spark.sql(f"""
    SELECT
        fecha,
        COUNT(*) AS transacciones,
        SUM(CAST(es_fraude AS INT)) AS transacciones_fraude,
        ROUND(SUM(CAST(es_fraude AS INT)) * 100.0 / COUNT(*), 2) AS tasa_fraude_pct
    FROM {tbl(SCHEMA_SILVER, "transacciones")}
    GROUP BY fecha
    ORDER BY fecha DESC
    LIMIT 7
""")
display(df_alerta)

UMBRAL_ALERTA_PCT = 5.0
ultimo_dia = df_alerta.first()
if ultimo_dia["tasa_fraude_pct"] > UMBRAL_ALERTA_PCT:
    print(f"[ALERTA] Tasa de fraude del {ultimo_dia['fecha']}: {ultimo_dia['tasa_fraude_pct']}% (umbral: {UMBRAL_ALERTA_PCT}%)")
else:
    print(f"[OK] Tasa de fraude del {ultimo_dia['fecha']}: {ultimo_dia['tasa_fraude_pct']}% — dentro del umbral")
````

**Como se crearia esta alerta en la interfaz** (Databricks SQL, sobre el SQL Warehouse 2X-Small de Free Edition):
1. Guardar el query anterior como un **Query** en Databricks SQL.
2. `SQL > Alerts > Create Alert` — condicion: `tasa_fraude_pct > 5.0` sobre la fila mas reciente.
3. Programar evaluacion cada hora; canal de notificacion: email del equipo de riesgo + webhook a Slack.
4. Adjuntar la Alert como una task de tipo *SQL Alert* dentro de este mismo Job, para que quede documentada junto con el pipeline que la alimenta.
