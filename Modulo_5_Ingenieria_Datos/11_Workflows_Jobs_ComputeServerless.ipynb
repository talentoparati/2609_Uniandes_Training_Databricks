Hasta ahora, cada notebook se ejecuto manualmente, celda por celda. En produccion, alguien tiene que correr estos mismos notebooks **todos los dias, en el orden correcto, con reintentos si algo falla**. Eso es un **Job** (Workflow):

- **Task**: la unidad minima de trabajo — normalmente un notebook, a veces un pipeline declarativo, un script Python o un query SQL.
- **Dependencia** (`depends_on`): un task no arranca hasta que sus predecesores terminan exitosamente.
- **Trigger**: que dispara el Job — un horario (cron), la llegada de un archivo, u otro Job (encadenamiento).
- **Compute**: cada task puede correr en Compute Serverless para Jobs — sin cluster que aprovisionar ni apagar manualmente, facturado solo por el tiempo de ejecucion del task.

```python
try:
    from databricks.sdk import WorkspaceClient
    from databricks.sdk.service import jobs
    print("databricks-sdk disponible.")
except ImportError:
    print("databricks-sdk no encontrado — instalando...")
    # %pip install databricks-sdk -q   # descomentar si el import anterior falla
    raise

w = WorkspaceClient()
print(f"Workspace: {w.config.host}")
```

      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">dbutils.notebook.getContext().notebookPath()</code> devuelve la ruta del notebook que esta corriendo esta celda ahora mismo; se le quitan los ultimos 2 segmentos (carpeta del modulo + nombre del notebook) para llegar a la raiz del repositorio.</td>

```python
# Ruta base del repositorio en el workspace — se deriva del notebook actual,
# asi el codigo funciona sin importar donde este montado el Git folder
_current_path = dbutils.notebook.entry_point.getDbutils().notebook().getContext().notebookPath().get()
BASE_PATH = "/".join(_current_path.split("/")[:-2])
print(f"Repositorio del curso montado en: {BASE_PATH}")
```

      <td style="padding:8px 16px 6px 0;color:#e2e8f0;font-size:13px;">Crea un Job de 2 tasks encadenadas: <code style="color:#38bdf8;">validar_entorno</code> &rarr; <code style="color:#38bdf8;">transformar_bronze_silver</code>.</td>
    </tr>

      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;">Primer Workflow real del curso: automatizar, con dependencias explicitas, lo que hasta ahora se ejecutaba a mano notebook por notebook.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">w.jobs.create()</code> recibe una lista de <code style="color:#38bdf8;">jobs.Task</code>; cada uno con un <code style="color:#38bdf8;">task_key</code> unico, un <code style="color:#38bdf8;">notebook_task</code> que apunta a la ruta absoluta calculada arriba, y opcionalmente <code style="color:#38bdf8;">depends_on</code>. Al no especificar <code style="color:#38bdf8;">cluster_spec</code> ni <code style="color:#38bdf8;">existing_cluster_id</code>, cada task corre automaticamente en <strong style="color:#e2e8f0;">Compute Serverless para Jobs</strong> — el comportamiento por defecto (y unico) en Free Edition.</td>
    </tr>
  </table>
</div>


```python
# Job de 2 tasks: validar entorno -> transformar Bronze -> Silver (Modulo 2)
# Sin cluster_spec => cada task corre en Compute Serverless para Jobs

nuevo_job = w.jobs.create(
    name=f"[{ENTORNO}] ecommerce_bronze_a_silver",
    tasks=[
        jobs.Task(
            task_key="validar_entorno",
            notebook_task=jobs.NotebookTask(notebook_path=f"{BASE_PATH}/00_Setup/00_Configuracion"),
        ),
        jobs.Task(
            task_key="transformar_bronze_silver",
            depends_on=[jobs.TaskDependency(task_key="validar_entorno")],
            notebook_task=jobs.NotebookTask(
                notebook_path=f"{BASE_PATH}/Modulo_2_Procesamiento_Spark/03_Transformacion_DataFrame_SparkSQL.ipynb"
            ),
        ),
    ],
    tags={"curso": "uniandes-databricks", "modulo": "5"},
)

print(f"[OK] Job creado — job_id = {nuevo_job.job_id}")
print(f"URL: {w.config.host}/jobs/{nuevo_job.job_id}")
```
