Hasta ahora, cada notebook calculaba sus propias columnas derivadas desde cero. Una **Feature Table** en Unity Catalog resuelve un problema muy concreto: evitar que el equipo de ML recalcule (y posiblemente reimplemente distinto) las mismas features que ya calculo el equipo de datos. Se publica **una vez**, con una *primary key*, y cualquier notebook de entrenamiento la reutiliza con un simple *lookup*.

```python
try:
    from databricks.feature_engineering import FeatureEngineeringClient, FeatureLookup
except ImportError:
    print("databricks-feature-engineering no encontrado — ejecute en una celda aparte:")
    print("  %pip install databricks-feature-engineering -q")
    print("  dbutils.library.restartPython()")
    print("y luego vuelva a correr %run ../00_Setup/00_Configuracion antes de continuar.")
    raise

fe = FeatureEngineeringClient()
FEATURE_TABLE = tbl(SCHEMA_GOLD, "features_riesgo_cliente")

df_features = spark.table(tbl(SCHEMA_GOLD, "clientes_riesgo_fraude"))

try:
    fe.create_table(
        name=FEATURE_TABLE,
        primary_keys=["cliente_id"],
        df=df_features,
        description="Features de riesgo de fraude por cliente — agregadas desde silver.transacciones (Modulo 5, Sesion 12)",
    )
    print(f"[OK] Feature table creada: {FEATURE_TABLE}")
except Exception as e:
    print(f"Feature table ya existia — actualizando con write_table (mode='merge')")
    fe.write_table(name=FEATURE_TABLE, df=df_features, mode="merge")

display(spark.table(FEATURE_TABLE).limit(5))
```

      <td style="padding:8px 16px 6px 0;color:#e2e8f0;font-size:13px;">Construye el set de entrenamiento a nivel transaccion usando <code style="color:#38bdf8;">FeatureLookup</code> para traer, por <code style="color:#38bdf8;">cliente_id</code>, las features de la feature table publicada.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">PARA QUE</td>
      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;">La etiqueta <code style="color:#38bdf8;">es_fraude</code> vive en <code style="color:#38bdf8;">silver.transacciones</code>, no en la feature table — por eso el set de entrenamiento parte de las transacciones y les <em>agrega</em> las features del cliente, sin volver a escribir el <code style="color:#38bdf8;">groupBy</code> del Modulo 5.</td>

      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">FeatureLookup(table_name, lookup_key, feature_names)</code> declara que columnas traer y por que clave unirlas; <code style="color:#38bdf8;">fe.create_training_set()</code> ejecuta ese join y ademas deja registrado en MLflow <em>de que feature table salio cada columna</em> — lo que mas adelante permite reconstruir exactamente los mismos features en produccion (Sesion 16, Model Serving). <code style="color:#38bdf8;">toPandas()</code> trae el resultado final a un DataFrame de pandas para entrenar con scikit-learn.</td>
    </tr>
  </table>
</div>


```python
df_labels = (
    spark.table(tbl(SCHEMA_SILVER, "transacciones"))
    .select("transaccion_id", "cliente_id", "monto_total", "metodo_pago", "dispositivo",
            "pais_cliente", "pais_ip_transaccion", "es_fraude")
    .withColumn("pais_distinto", (F.col("pais_ip_transaccion") != F.col("pais_cliente")).cast("int"))
    .drop("pais_cliente", "pais_ip_transaccion")
)

feature_lookups = [
    FeatureLookup(
        table_name=FEATURE_TABLE,
        lookup_key="cliente_id",
        feature_names=["num_transacciones", "tasa_fraude_historica", "monto_promedio", "monto_maximo", "num_pais_distinto", "segmento"],
    )
]

training_set = fe.create_training_set(
    df=df_labels,
    feature_lookups=feature_lookups,
    label="es_fraude",
    exclude_columns=["transaccion_id", "cliente_id"],
)

pdf = training_set.load_df().toPandas()
print(f"Filas de entrenamiento: {len(pdf):,}")
pdf.head()
```

```python
import mlflow
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score, classification_report, confusion_matrix

mlflow.set_experiment(f"/Shared/{ENTORNO}_deteccion_fraude")

pdf_encoded = pd.get_dummies(pdf, columns=["metodo_pago", "dispositivo", "segmento"], drop_first=True)
X = pdf_encoded.drop(columns=["es_fraude"])
y = pdf_encoded["es_fraude"].astype(int)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=SEED, stratify=y)
print(f"Train: {len(X_train):,}  |  Test: {len(X_test):,}  |  Tasa de fraude en train: {y_train.mean()*100:.2f}%")
```

```python
with mlflow.start_run(run_name="random_forest_fraude") as run:
    params = {"n_estimators": 200, "max_depth": 8, "class_weight": "balanced", "random_state": SEED}
    mlflow.log_params(params)

    modelo = RandomForestClassifier(**params)
    modelo.fit(X_train, y_train)

    proba_test = modelo.predict_proba(X_test)[:, 1]
    pred_test = modelo.predict(X_test)
    auc = roc_auc_score(y_test, proba_test)

    mlflow.log_metric("auc_test", auc)
    reporte = classification_report(y_test, pred_test, output_dict=True)
    mlflow.log_metric("precision_fraude", reporte["1"]["precision"])
    mlflow.log_metric("recall_fraude", reporte["1"]["recall"])

    fe.log_model(
        model=modelo,
        artifact_path="modelo",
        flavor=mlflow.sklearn,
        training_set=training_set,
        registered_model_name=None,  # se registra explicitamente mas abajo
    )

    run_id = run.info.run_id

print(f"AUC (test): {auc:.4f}")
print(f"Precision (clase fraude): {reporte['1']['precision']:.3f}")
print(f"Recall (clase fraude)   : {reporte['1']['recall']:.3f}")
print(f"\nMatriz de confusion:\n{confusion_matrix(y_test, pred_test)}")
print(f"\nRun registrado: {run_id}")
```

    <tr>
      <td style="width:78px;padding:8px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">QUE HACE</td>
      <td style="padding:8px 16px 6px 0;color:#e2e8f0;font-size:13px;">Entrena el mismo problema de clasificacion con <code style="color:#38bdf8;">pyspark.ml</code> (Spark ML) en vez de scikit-learn, dentro de otro run de MLflow del mismo experimento.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">PARA QUE</td>
      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">scikit-learn</code> exige traer todos los datos a un solo proceso (<code style="color:#38bdf8;">toPandas()</code>) — razonable para los volumenes de este curso, pero inviable si el set de entrenamiento no cabe en un solo nodo. <strong style="color:#e2e8f0;">Spark ML</strong> resuelve el mismo problema de forma distribuida.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">StringIndexer</code> convierte texto a indices numericos, <code style="color:#38bdf8;">VectorAssembler</code> combina todas las columnas en un solo vector <code style="color:#38bdf8;">features</code>, y <code style="color:#38bdf8;">Pipeline</code> encadena ambos pasos con el clasificador — la misma API declarativa (<code style="color:#38bdf8;">Pipeline</code>) que ya se vio conceptualmente en Spark SQL. <code style="color:#38bdf8;">mlflow.spark.log_model()</code> guarda el pipeline completo (transformaciones + modelo) como un unico artefacto.</td>
    </tr>
  </table>
</div>


```python
from pyspark.ml.feature import StringIndexer, VectorAssembler
from pyspark.ml.classification import RandomForestClassifier as SparkRF
from pyspark.ml import Pipeline
from pyspark.ml.evaluation import BinaryClassificationEvaluator

df_spark_ml = training_set.load_df().withColumn("label", F.col("es_fraude").cast("double"))
train_sp, test_sp = df_spark_ml.randomSplit([0.8, 0.2], seed=SEED)

indexers = [StringIndexer(inputCol=c, outputCol=f"{c}_idx", handleInvalid="keep")
            for c in ["metodo_pago", "dispositivo", "segmento"]]
ensamblador = VectorAssembler(
    inputCols=["monto_total", "pais_distinto", "num_transacciones", "tasa_fraude_historica",
               "monto_promedio", "monto_maximo", "num_pais_distinto"] + [f"{c}_idx" for c in ["metodo_pago", "dispositivo", "segmento"]],
    outputCol="features",
)
clasificador = SparkRF(featuresCol="features", labelCol="label", numTrees=200, maxDepth=8, seed=SEED)

pipeline_ml = Pipeline(stages=indexers + [ensamblador, clasificador])

with mlflow.start_run(run_name="spark_ml_fraude"):
    modelo_spark = pipeline_ml.fit(train_sp)
    predicciones = modelo_spark.transform(test_sp)

    evaluador = BinaryClassificationEvaluator(labelCol="label", metricName="areaUnderROC")
    auc_spark = evaluador.evaluate(predicciones)
    mlflow.log_metric("auc_test", auc_spark)
    mlflow.spark.log_model(modelo_spark, artifact_path="modelo_spark")

print(f"AUC (Spark ML, test): {auc_spark:.4f}")
print("Ambos runs (sklearn y Spark ML) quedan visibles y comparables en el mismo experimento de MLflow.")
```
