      <td style="padding:8px 16px 6px 0;color:#e2e8f0;font-size:13px;">Lista todas las versiones registradas de <code style="color:#38bdf8;">MODEL_NAME</code> y asigna el alias <code style="color:#38bdf8;">champion</code> a la mas reciente.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">PARA QUE</td>
      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;">El Modulo 6 (Sesion 14) registro una version del modelo, pero <strong style="color:#e2e8f0;">registrado</strong> no es lo mismo que <strong style="color:#e2e8f0;">en produccion</strong>. Unity Catalog usa <strong>alias</strong> (por ejemplo <code style="color:#38bdf8;">champion</code>) en vez de los antiguos "Staging/Production" de MLflow clasico: un alias apunta a una version especifica y se puede <em>mover</em> instantaneamente a una version nueva sin cambiar el codigo que consume el modelo.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">MlflowClient().search_model_versions()</code> lista versiones con su <code style="color:#38bdf8;">run_id</code> de origen; <code style="color:#38bdf8;">set_registered_model_alias()</code> mueve el alias a la version indicada. El consumidor final siempre pide <code style="color:#38bdf8;">models:/{MODEL_NAME}@champion</code>, nunca un numero de version fijo.</td>
    </tr>
  </table>
</div>


```python
from mlflow import MlflowClient

client = MlflowClient()
versiones = client.search_model_versions(f"name = '{MODEL_NAME}'")
print(f"Versiones registradas de {MODEL_NAME}:")
for v in versiones:
    print(f"  version {v.version}  |  run_id={v.run_id}  |  status={v.status}")

ultima_version = max(int(v.version) for v in versiones)

client.set_registered_model_alias(name=MODEL_NAME, alias="champion", version=ultima_version)
print(f"\n[OK] Alias 'champion' -> version {ultima_version}")
```

  <span style="color:#cbd5e1;font-size:13px;">Databricks Free Edition incluye <strong style="color:#e2e8f0;">Model Serving</strong> con restricciones claras: sin serving con GPU, sin <em>provisioned throughput</em>, y un numero limitado de endpoints activos. Un modelo scikit-learn como el de este curso no necesita GPU, asi que corre sin problema dentro de esos limites con <code style="color:#38bdf8;">workload_size="Small"</code> y <code style="color:#38bdf8;">scale_to_zero_enabled=True</code> (el endpoint se apaga solo sin trafico, controlando el costo/cuota).</span>

      <td style="padding:8px 16px 6px 0;color:#e2e8f0;font-size:13px;">Crea un endpoint de Model Serving (CPU, tamano Small, con scale-to-zero) que sirve la ultima version registrada del modelo.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">PARA QUE</td>
      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;">Empaquetar el modelo (con todas sus dependencias, capturadas automaticamente por MLflow al momento de <code style="color:#38bdf8;">log_model</code>) detras de un endpoint REST administrado — sin que el equipo tenga que construir un contenedor ni una API a mano.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">w.serving_endpoints.create()</code> recibe un <code style="color:#38bdf8;">EndpointCoreConfigInput</code> con uno o mas <code style="color:#38bdf8;">ServedEntityInput</code> — aqui, uno solo, apuntando a <code style="color:#38bdf8;">entity_name</code>/<code style="color:#38bdf8;">entity_version</code> del modelo recien registrado. El <code style="color:#38bdf8;">try/except</code> cubre el caso de volver a ejecutar la celda cuando el endpoint ya existe.</td>
    </tr>
  </table>
</div>


```python
from databricks.sdk import WorkspaceClient
from databricks.sdk.service.serving import EndpointCoreConfigInput, ServedEntityInput

w = WorkspaceClient()
ENDPOINT_NAME = f"{ENTORNO}-deteccion-fraude"

try:
    endpoint = w.serving_endpoints.create(
        name=ENDPOINT_NAME,
        config=EndpointCoreConfigInput(
            served_entities=[
                ServedEntityInput(
                    entity_name=MODEL_NAME,
                    entity_version=str(ultima_version),
                    workload_size="Small",
                    scale_to_zero_enabled=True,  # CPU, escala a 0 cuando no hay trafico — controla costo/cuota
                )
            ]
        ),
    )
    print(f"[OK] Endpoint '{ENDPOINT_NAME}' creado — aprovisionando (puede tardar varios minutos)")
except Exception as e:
    print(f"El endpoint ya existe o esta aprovisionandose: {str(e)[:200]}")

print(f"\nURL del endpoint (una vez listo):")
print(f"  {w.config.host}/serving-endpoints/{ENDPOINT_NAME}/invocations")
```
