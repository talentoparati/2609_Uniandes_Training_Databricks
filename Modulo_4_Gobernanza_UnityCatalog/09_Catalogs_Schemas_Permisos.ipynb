  <h2 style="color:#ffffff;font-size:20px;margin:4px 0 0 0;">Tablas administradas (managed) vs externas</h2>
</div>


Todas las tablas creadas hasta ahora en el curso son **managed tables**: Unity Catalog controla tanto el metadato como los archivos fisicos. Si se hace `DROP TABLE`, los archivos se borran tambien. Una **external table** solo registra el metadato — los archivos viven en una ubicacion externa (por ejemplo un bucket que tambien usan otros sistemas) y sobreviven a un `DROP TABLE`.

| Criterio | Managed | External |
|---|---|---|
| Duenos de los archivos | Unity Catalog | Sistema externo |
| `DROP TABLE` | Borra metadato **y** archivos | Borra solo metadato |
| Ubicacion | Definida por el catalog/schema | Definida por el usuario (`LOCATION`) |
| Caso de uso tipico | Datos que "viven" en el Lakehouse | Datos compartidos con otro sistema o legados de migracion |

<div style="box-sizing:border-box;width:100%;background:#0c1a2e;border-left:4px solid #f59e0b;padding:14px 16px;border-radius:6px;margin:10px 0;">
  <span style="color:#f59e0b;font-size:11px;font-weight:700;letter-spacing:1px;">DATABRICKS FREE EDITION — WORKSPACE DE UN SOLO USUARIO</span><br/>
  <span style="color:#cbd5e1;font-size:13px;">Databricks Free Edition es un workspace <strong style="color:#e2e8f0;">de un solo usuario</strong>: no tiene consola de cuenta ni aprovisionamiento de grupos (SCIM). Por eso, en las celdas siguientes, el <code style="color:#38bdf8;">GRANT</code>/<code style="color:#38bdf8;">REVOKE</code> se aplica sobre <code style="color:#38bdf8;">current_user()</code> — la propia cuenta — como <em>stand-in</em> de un grupo real. La sintaxis es identica a la que se usaria en un workspace de equipo con grupos como <code style="color:#38bdf8;">analistas_bi</code> o <code style="color:#38bdf8;">ingenieros_datos</code>; solo cambia el nombre del principal.</span>
</div>


<div style="box-sizing:border-box;width:100%;background:#0c1a2e;border:1px solid #1e3a5f;border-left:4px solid #2563eb;border-radius:6px;margin:10px 0;overflow:hidden;">

    <tr>
      <td style="width:78px;padding:8px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">QUE HACE</td>
      <td style="padding:8px 16px 6px 0;color:#e2e8f0;font-size:13px;">Otorga <code style="color:#38bdf8;">USE CATALOG</code>, <code style="color:#38bdf8;">USE SCHEMA</code> y <code style="color:#38bdf8;">SELECT</code> sobre el schema Gold al usuario actual (stand-in de un grupo de BI).</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">PARA QUE</td>
      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;">Unity Catalog aplica permisos <strong style="color:#e2e8f0;">jerarquicos</strong>: para leer una tabla, un principal necesita los tres permisos a la vez. El escenario tipico de este curso: un equipo de <strong style="color:#e2e8f0;">BI/Analitica</strong> solo deberia ver la capa Gold (KPIs ya agregados), nunca Bronze (datos crudos con PII como email) — el <em>principio de minimo privilegio</em>.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;"><code style="color:#38bdf8;">GRANT privilegio ON objeto TO principal</code> es DDL estandar de Unity Catalog. <code style="color:#38bdf8;">current_user()</code> devuelve el email/identidad de la sesion actual — en un workspace de equipo, ese mismo string se reemplaza por el nombre de un grupo.</td>
    </tr>
  </table>
</div>


```python
# Usuario/principal actual — en un workspace de equipo se otorgaria a un grupo (p.ej. "analistas_bi")
usuario_actual = spark.sql("SELECT current_user() AS u").first()["u"]
print(f"Principal actual: {usuario_actual}")

# Otorgar acceso de solo lectura a Gold (analistas BI) — aqui, el usuario actual como stand-in de un grupo
for stmt in [
    f"GRANT USE CATALOG ON CATALOG {CATALOG_NAME} TO `{usuario_actual}`",
    f"GRANT USE SCHEMA ON SCHEMA {CATALOG_NAME}.{SCHEMA_GOLD} TO `{usuario_actual}`",
    f"GRANT SELECT ON SCHEMA {CATALOG_NAME}.{SCHEMA_GOLD} TO `{usuario_actual}`",
]:
    spark.sql(stmt)
    print(f"[OK] {stmt}")

print("\nEn un workspace de equipo, con grupos configurados, se reemplaza el usuario por el grupo:")
print(f"  GRANT SELECT ON SCHEMA {CATALOG_NAME}.{SCHEMA_GOLD} TO `analistas_bi`")
print(f"  GRANT ALL PRIVILEGES ON SCHEMA {CATALOG_NAME}.{SCHEMA_BRONZE} TO `ingenieros_datos`")
```
