  <span style="color:#0d9488;font-size:11px;font-weight:700;letter-spacing:1px;">CATALOG EXPLORER — LINEAGE</span><br/>
  <span style="color:#cbd5e1;font-size:13px;">El <strong style="color:#e2e8f0;">Catalog Explorer</strong> tiene una pestana <strong>Lineage</strong> en cada tabla: muestra, de forma grafica, que tablas/notebooks/Jobs leyeron y escribieron esa tabla — sin que nadie haya instrumentado nada. Unity Catalog lo construye automaticamente a partir de cada consulta ejecutada. Por ejemplo, <code style="color:#38bdf8;">silver.transacciones</code> deberia mostrar como upstream a <code style="color:#38bdf8;">bronze.transacciones</code>, y como notebooks que la escribieron, los de la Sesion 6. Esta pestana visual funciona en Free Edition sin configuracion adicional.</span>
</div>


<div style="box-sizing:border-box;width:100%;background:#0c1a2e;border:1px solid #1e3a5f;border-left:4px solid #2563eb;border-radius:6px;margin:10px 0;overflow:hidden;">
  <div style="padding:8px 16px;border-bottom:1px solid #1e3a5f;background:rgba(37,99,235,0.08);">
    <span style="color:#38bdf8;font-size:10px;font-weight:700;letter-spacing:1.5px;">ANTES DE EJECUTAR ESTA CELDA</span>
  </div>
  <table style="width:100%;border-collapse:collapse;">
    <tr>
      <td style="width:78px;padding:8px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">QUE HACE</td>
      <td style="padding:8px 16px 6px 0;color:#e2e8f0;font-size:13px;">Intenta consultar <code style="color:#38bdf8;">system.access.table_lineage</code> por SQL — el mismo linaje que muestra el Catalog Explorer, pero consultable.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">PARA QUE</td>
      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;">Util para auditorias automatizadas (por ejemplo, "que tablas Gold dependen de una tabla Bronze que vamos a deprecar") en vez de revisar tabla por tabla en el UI.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;">Las <strong style="color:#e2e8f0;">System Tables</strong> de linaje/auditoria requieren habilitacion a nivel de cuenta y pueden no estar disponibles en todos los workspaces de Free Edition — por eso la consulta va envuelta en <code style="color:#38bdf8;">try/except</code>: si falla o no devuelve filas, el codigo lo explica en vez de romper el notebook, y recuerda que la pestana visual de Lineage siempre esta disponible como alternativa.</td>
    </tr>
  </table>
</div>


```python
try:
    df_lineage = spark.sql(f"""
        SELECT source_table_full_name, target_table_full_name, source_type, target_type
        FROM system.access.table_lineage
        WHERE target_table_full_name LIKE '{CATALOG_NAME}.%'
        ORDER BY target_table_full_name
        LIMIT 50
    """)
    if df_lineage.count() == 0:
        print("Sin filas — System Tables probablemente no estan habilitadas en este workspace.")
        print("Habilitar en: Account Console > Settings > System Tables (requiere admin de cuenta).")
    else:
        display(df_lineage)
except Exception as e:
    print("system.access.table_lineage no disponible en este workspace:")
    print(f"  {str(e)[:200]}")
    print("\nAlternativa disponible siempre: pestana 'Lineage' en Catalog Explorer, sobre cada tabla.")
```

      <td style="padding:8px 16px 6px 0;color:#e2e8f0;font-size:13px;">Intenta consultar <code style="color:#38bdf8;">system.access.audit</code> para ver los ultimos accesos registrados sobre objetos de Unity Catalog.</td>

    <tr>
      <td style="padding:6px 4px 6px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">PARA QUE</td>
      <td style="padding:6px 16px 6px 0;color:#93c5fd;font-size:13px;">Es la fuente de verdad para responder preguntas como "¿quien consulto la tabla de transacciones (con datos de pago) la semana pasada?" — exactamente el tipo de pregunta que un area de seguridad o cumplimiento hace despues de un incidente.</td>
    </tr>
    <tr>
      <td style="padding:6px 4px 10px 16px;vertical-align:top;color:#64748b;font-size:10px;letter-spacing:1px;white-space:nowrap;">COMO LO HACE</td>
      <td style="padding:6px 16px 10px 0;color:#93c5fd;font-size:13px;">Mismo patron defensivo que la celda de linaje: <code style="color:#38bdf8;">try/except</code> porque <code style="color:#38bdf8;">system.access.audit</code> tambien depende de que las System Tables esten habilitadas para esta cuenta.</td>
    </tr>
  </table>
</div>


```python
try:
    df_audit = spark.sql(f"""
        SELECT event_time, user_identity.email AS usuario, action_name, request_params.full_name_arg AS objeto
        FROM system.access.audit
        WHERE service_name = 'unityCatalog'
          AND action_name IN ('getTable', 'createTable', 'updateTable')
        ORDER BY event_time DESC
        LIMIT 20
    """)
    if df_audit.count() == 0:
        print("Sin filas — System Tables probablemente no estan habilitadas, o aun no hay eventos recientes.")
    else:
        display(df_audit)
except Exception as e:
    print("system.access.audit no disponible en este workspace:")
    print(f"  {str(e)[:200]}")
    print("\nEsta tabla es el equivalente Unity Catalog a los audit logs de un SGBD tradicional.")
```
